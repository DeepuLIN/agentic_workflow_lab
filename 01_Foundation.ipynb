```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
GROQ_API_KEY = os.getenv('GROQ_API_KEY')
GOOGLE_API_KEY = os.getenv('GOOGLE_API_KEY')
```

```python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=GOOGLE_API_KEY)
ollama = OpenAI(api_key="ollama", base_url="http://localhost:11434/v1")
```

```python
message = [
    {"role":"user", "content" :"What is 2+2"}
]
```

```python
response = gemini.chat.completions.create(model="gemini-2.5-flash-lite", messages=message)
display(Markdown(response.choices[0].message.content))
```

```python
response = ollama.chat.completions.create(messages=message, model="llama3.2:latest")
display(Markdown(response.choices[0].message.content))
```

```python
question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
response = ollama.chat.completions.create(messages=messages, model="llama3.2:latest")
display(Markdown(response.choices[0].message.content))
```

```python
response = gemini.chat.completions.create(messages=messages, model="gemini-2.5-flash-lite")
question = response.choices[0].message.content
display(Markdown(response.choices[0].message.content))
```

```python
messages = [{"role": "user", "content": question}]
answer = gemini.chat.completions.create(messages=messages, model="gemini-2.5-flash-lite")
display(Markdown(answer.choices[0].message.content))
```
